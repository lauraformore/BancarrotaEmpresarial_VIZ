## 1. INTRODUCCIÓN

La quiebra o bancarrota empresarial es un evento crítico con profundas repercusiones para accionistas, acreedores, empleados y la economía en general. Identificar de manera temprana las señales de deterioro financiero permite a los tomadores de decisiones implementar medidas correctivas, gestionar el riesgo crediticio y optimizar portafolios de inversión.

Este proyecto aborda la problemática del riesgo de insolvencia utilizando un conjunto de datos del **Taiwan Economic Journal (1999–2009)**, compuesto por indicadores financieros de empresas que cotizan en la Bolsa de Taiwán. Mediante el análisis de 95 razones financieras (*ratios* de rentabilidad, liquidez, apalancamiento y eficiencia), el proyecto busca construir una solución integral que va desde el entendimiento de los datos hasta la implementación de una herramienta interactiva para la toma de decisiones.

Este proyecto abarca el **Análisis Exploratorio de Datos (EDA)**, el **modelado predictivo mediante Machine Learning** y la **creación de un Dashboard interactivo en Dash**, permitiendo así una evaluación de riesgo automatizada, precisa y visualmente accesible.

## 2. OBJETIVOS DEL PROYECTO

### Objetivo General
Desarrollar un sistema integral de evaluación de riesgo financiero mediante el análisis exploratorio de datos, el entrenamiento de modelos predictivos de Machine Learning para la detección temprana de bancarrota empresarial, y la implementación de un Dashboard interactivo en Dash para la visualización y consulta de resultados.

### Objetivos Específicos

1. **Análisis Exploratorio y Preparación de Datos (EDA & ETL):**
   * Limpiar, estructurar y validar la calidad de los datos financieros.
   * Analizar la distribución de los 95 ratios financieros y comprender el desbalance crítico en la variable objetivo (`Bankrupt?`).
   * Aplicar una división de datos estricta (`train`/`test`) mediante muestreo estratificado para prevenir el sesgo de información (*data leakage*).
   * Identificar las variables financieras con mayor poder discriminatorio entre empresas estables y propensas a la quiebra.

2. **Modelado Predictivo:**
   * Tratar el desbalance de clases mediante técnicas apropiadas de resuestreo o ajuste de pesos en los algoritmos.
   * Entrenar y evaluar algoritmos de clasificación (ej. Regresión Logística, Random Forest, XGBoost) orientados a maximizar métricas clave como *Recall*, *ROC-AUC* y *PR-AUC*.
   * Seleccionar e interpretar el mejor modelo predictivo, identificando la importancia de las variables en la predicción del riesgo de insolvencia.

3. **Desarrollo del Dashboard Interactivo (Dash):**
   * Diseñar una interfaz analítica en Dash que presente de forma clara los principales indicadores macro e individuales del dataset.
   * Integrar el modelo de Machine Learning dentro de la aplicación para permitir simulaciones en tiempo real del riesgo de bancarrota según los ratios financieros ingresados.
   * Desplegar la herramienta final facilitando la exploración interactiva para usuarios y tomadores de decisiones.
